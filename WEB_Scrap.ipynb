## Extract faculty details

### Subtask:
Iterate through each faculty div, extract the name and role. Then, parse the paragraph text to extract qualification, total experience, email, and phone number for each faculty member.


**Reasoning**:
The subtask requires extracting specific details (name, role, qualification, experience, email, and phone) for each faculty member from the parsed HTML. I will iterate through the relevant HTML divisions and use string manipulation to extract the information into separate lists.



```python
import requests

faculty_names = []
faculty_roles = []
faculty_qualifications = []
faculty_experiences = []
faculty_emails = []
faculty_phones = []

for faculty_div in soups.find_all('div', class_='class1'):
    # Skip the H1 tags that are not part of faculty details
    if faculty_div.find('h1'):
        continue

    # Extract name
    name_tag = faculty_div.find('h3')
    faculty_names.append(name_tag.text.strip() if name_tag else None)

    # Extract role
    role_tag = faculty_div.find('h4')
    faculty_roles.append(role_tag.text.strip() if role_tag else None)

    # Extract paragraph content for other details
    p_tag = faculty_div.find('p', class_='class5')
    if p_tag:
        p_text = p_tag.text.replace('\n', ' ').strip()

        # Extract Qualification
        qualification_match = re.search(r'Qualification:\s*(.*?)(?:\s*Total Experience:|$)', p_text)
        faculty_qualifications.append(qualification_match.group(1).strip() if qualification_match else None)

        # Extract Total Experience
        experience_match = re.search(r'Total Experience:\s*(.*?)(?:\s*Publications:|$)', p_text)
        faculty_experiences.append(experience_match.group(1).strip() if experience_match else None)

        # Extract Publications (if needed, but not asked explicitly)
        # publications_match = re.search(r'Publications:\s*(.*?)(?:\s*Email:|$)', p_text)
        # publications.append(publications_match.group(1).strip() if publications_match else None)

        # Extract Email
        email_match = re.search(r'Email:\s*(.*?)(?:\s*Phone:|$)', p_text)
        faculty_emails.append(email_match.group(1).strip() if email_match else None)

        # Extract Phone
        phone_match = re.search(r'Phone:\s*(.*)', p_text)
        faculty_phones.append(phone_match.group(1).strip() if phone_match else None)
    else:
        faculty_qualifications.append(None)
        faculty_experiences.append(None)
        faculty_emails.append(None)
        faculty_phones.append(None)

print("Extraction complete. Data stored in respective lists.")
```

The faculty information has been successfully extracted into separate lists. Now, to display this data in a structured and organized format, I will combine these lists into a pandas DataFrame.
